```python
gdf = gpd.read_file("dataset/raw_data/indexes/ndvi_mean.geojson")
gdf['value'] = gdf['value'].astype(float)
print(gdf.value.max())
print(gdf.value.min())
print(gdf.value.quantile(0.05))
print(gdf.value.quantile(0.95))
```

```python
gdf = gpd.read_file("dataset/raw_data/indexes/urban_heat_exposure_index.geojson")
gdf['value'] = gdf['value'].astype(float)
gdf['inverse_uhei'] = 2.815929651260376 - gdf['value']

print("5th percentile:", gdf['inverse_uhei'].quantile(0.05))
print("95th percentile:", gdf['inverse_uhei'].quantile(0.95))
print("Min:", gdf['inverse_uhei'].min())
print("Max:", gdf['inverse_uhei'].max())
```
